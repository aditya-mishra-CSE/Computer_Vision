```python
import pandas as pd
```

```python
data = {
    'age': [25,35,45,23,54,28,32,42,37],
    'salary': [5000,6000,10000,4000,12000,35000,110000,6500,9500],
    'Purchase Amount': [200,500,700,150,800,300,750,400,650]
}

df = pd.DataFrame(data)
df
```

```python
df.describe()
```

```python
#Normalizing the dataset
normalized_df = (df-df.min())/(df.max()-df.min())
normalized_df
```

```python
normalized_df.describe()
```

```python
#Standardizing the dataset
standardized_df = (df-df.mean())/df.std()
standardized_df
```

```python
standardized_df.describe()
```
